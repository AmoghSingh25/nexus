```python
import os
import numpy as np
import pandas as pd
import networkx as nx
import random
import matplotlib.pyplot as plt


%load_ext autoreload
%autoreload 2
```

```python
from simulator import gpsim
```

```python
sim = gpsim.simulator()
```

```python
r_vals, p_vals = [], []
no_steps = 10
for _ in range(no_steps):
    a,b = sim.ret_vals()
    r_vals.append(a)
    p_vals.append(b)
    for i in range(len(sim.g.nodes)):
        sim.calc_x_t(i)

p_vals = np.array(p_vals)
r_vals = np.array(r_vals)
```

```python
for i in range(p_vals.shape[1]):
    plt.plot(p_vals[:, i], label=f'Node {[i]}')

plt.title("Expression values of a single cell")
plt.xlabel('Steps')
plt.ylabel('Expression value')
plt.legend()
plt.grid(True)
plt.show()
```
